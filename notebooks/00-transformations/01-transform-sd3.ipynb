```python
import json
import random

random.seed(42)
```

```python
data = json.load(open("../../data/SD3_questions_paraphrased.json"))
```

```python
items = []
i = 0
for key, values in data.items():
    item = {
        "item_id": f"SD3{i}_V0",
        "text": key,
        "semantic_anchors": ["Strongly Disagree", "Disagree", "Neutral", "Agree", "Strongly Agree"]
    }
    items.append(item)

    # Sample 4 distinct paraphrases without replacement (excluding the original wording)
    candidates = [v for v in dict.fromkeys(values) if v != key]
    for j, value in enumerate(random.sample(candidates, k=4)):
        item = {
            "item_id": f"SD3{i}_V{j+1}",
            "text": value,
            "semantic_anchors": ["Strongly Disagree", "Disagree", "Neutral", "Agree", "Strongly Agree"]
        }
        items.append(item)
    i += 1
```

```python
with open("../../data/questionnaires/SD3_paraphrased.json", "w") as f:
    questionnaire = {
        "questionnaire_name": "sd3-with-paraphrases",
        "items": items
    }
    json.dump(questionnaire, f, indent=4)
```
